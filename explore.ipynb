## 4. Parse JSON body & flatten into DataFrame

```python
import json

# Parse one row to see the structure
raw = df.iloc[0]["body"]
data = json.loads(json.loads(raw))

print("Top-level keys:", list(data.keys()))
print("End point:", data.get("end_point"))
print("Method:", data.get("method"))
print("\nData keys:", list(data.get("data", {}).keys()))
```

```python
# Parse all rows and flatten the nested JSON into columns
def parse_body(raw):
    try:
        parsed = json.loads(json.loads(raw))
        result = {"end_point": parsed.get("end_point"), "method": parsed.get("method")}
        result.update(parsed.get("data", {}))
        return result
    except (json.JSONDecodeError, TypeError):
        return {}

parsed = df["body"].apply(parse_body)
df_body = pd.json_normalize(parsed)

print(f"Parsed rows: {len(df_body)}, Columns: {len(df_body.columns)}")
df_body.columns.tolist()
```

```python
# Pick useful columns to preview (this is just an example, adjust based on actual columns)
cols = [
    "id","end_point", "method", "number", "date", "description", "status",
    "supplier.name", "subtotal", "total_amount", "total_tax",
    "currency.code", "department.name", "project.name",
]
existing = [c for c in cols if c in df_body.columns]
df_body[existing].head(10)
```

## 5. Analyze data for pipeline dedup strategy

Load all data to understand: what end_point types exist, how many duplicates per transaction, and how to pick the latest status.

```python
# Load last 1 month of data (just id, body, created_at to keep it light)
df_all = pd.read_sql(
    f"SELECT id AS callback_id, body, created_at FROM {TABLE_NAME} "
    f"WHERE created_at >= DATE_SUB(NOW(), INTERVAL 6 MONTH) "
    f"ORDER BY created_at DESC",
    engine,
)
print(f"Total callbacks (last 1 month): {len(df_all)}")

# Parse all bodies
def parse_body_light(raw):
    try:
        parsed = json.loads(json.loads(raw))
        d = parsed.get("data", {})
        return {
            "end_point": parsed.get("end_point"),
            "method": parsed.get("method"),
            "txn_id": d.get("id"),
            "number": d.get("number"),
            "status": d.get("status"),
            "payment_status": d.get("payment_status"),
            "date": d.get("date"),
        }
    except (json.JSONDecodeError, TypeError):
        return {}

df_parsed = pd.DataFrame(df_all["created_at"])
df_parsed = df_parsed.join(pd.json_normalize(df_all["body"].apply(parse_body_light)))
print(f"Parsed: {len(df_parsed)} rows")
df_parsed.head()
```

```python
# What end_point types exist?
print("=== End Point Distribution ===")
print(df_parsed["end_point"].value_counts())
print()

# What methods (POST/PUT/PATCH)?
print("=== Method Distribution ===")
print(df_parsed["method"].value_counts())
print()

# What statuses?
print("=== Status Distribution ===")
print(df_parsed["status"].value_counts())
print()

# Draft vs Approved per end_point
print("=== Status by End Point ===")
print(pd.crosstab(df_parsed["end_point"], df_parsed["status"], margins=True))
```

```python
# How many callbacks per unique transaction?
# A transaction is identified by (end_point, txn_id)
dupes = df_parsed.groupby(["end_point", "txn_id"]).size().reset_index(name="callback_count")

print(f"Total callbacks: {len(df_parsed)}")
print(f"Unique transactions: {len(dupes)}")
print(f"Duplicated transactions: {(dupes['callback_count'] > 1).sum()}")
print()
print("=== Callbacks per transaction ===")
print(dupes["callback_count"].value_counts().sort_index())
```

```python
# Example: look at a duplicated transaction to see how status changes over time
duped_txns = dupes[dupes["callback_count"] > 1]
if not duped_txns.empty:
    sample_ep = duped_txns.iloc[0]["end_point"]
    sample_id = duped_txns.iloc[0]["txn_id"]

    sample = df_parsed[
        (df_parsed["end_point"] == sample_ep) & (df_parsed["txn_id"] == sample_id)
    ].sort_values("created_at")

    print(f"Transaction: {sample_ep} / {sample_id}")
    print(sample[["created_at", "method", "number", "status", "payment_status"]])
else:
    print("No duplicated transactions found")
```
